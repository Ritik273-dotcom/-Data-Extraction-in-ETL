Question 1 : Describe different types of data sources used in ETL with suitable examples.

Answer: The different types of data sources used in ETL (Extract, Transform, Load) processes are generally categorized by how the data is organized.

1. Structured Data SourcesThese sources follow a strict, predefined schema or format, making them highly predictable and easy to query.

Relational Databases (RDBMS): The most common source. Data is stored in tables with rows and columns.

Example: MySQL, Oracle, SQL Server, or PostgreSQL.

ERP/CRM Systems: Enterprise applications that store business-critical data in structured backends.

Example: SAP, Salesforce, or Microsoft Dynamics.

2. Semi-Structured Data Sources

These sources do not have a rigid tabular structure but contain tags or markers to separate data elements and enforce a hierarchy.

Web-based Formats: Commonly used for data exchange between applications and APIs.

Example: JSON (JavaScript Object Notation) or XML (eXtensible Markup Language) files.

Flat Files: Text files where data is separated by specific characters.

Example: CSV (Comma-Separated Values) or TSV (Tab-Separated Values) files.

3. Unstructured Data Sources

These sources have no predefined data model or organization, making them the most complex to process during the "Transform" phase.

Media and Documents: Content that requires specialized processing to extract meaningful data.

Example: PDFs, Images, Videos, Emails, or Social Media posts.

Log Files: Raw text generated by servers or applications.

Example: Web server logs or Application error logs.

4. Streaming Data Sources

Data that is generated continuously and must be processed in real-time or near-real-time.

IoT and Sensors: Continuous data feeds from hardware devices.

Example: Smart meter readings or GPS tracking data.

Message Queues: Real-time data streams used for high-frequency events.

Example: Apache Kafka or Amazon Kinesis.

Question 2 : What is data extraction? Explain its role in the ETL pipeline.

Answer: Data extraction is the first step in the data integration process, involving the retrieval of raw data from multiple, often disparate, source systems (like CRMs, ERPs, flat files, or relational databases) to be consolidated into a centralized repository.

Its Role in the ETL Pipeline

As the foundational block of an ETL (Extract, Transform, Load) pipeline, the extraction phase plays several critical roles:

Source Data Gathering: It locates and retrieves data from diverse sources without altering its original format.

Enabling Transformation: By successfully pulling data out of its isolated silos, it acts as a staging bridge so the data can be cleaned, structured, and reformatted in the next phase.

Controlling Pipeline Efficiency: Extraction methods directly impact pipeline performance.

Full Extraction: Pulls the entire dataset, commonly used for initial data loads.

Incremental Extraction: Pulls only newly added or modified data, which significantly reduces network load.

Question 3 : Explain the difference between CSV and Excel in terms of extraction and ETL usage.

Answer: CSV and Excel files differ fundamentally in their structure, causing major variations in how data pipelines extract and process them.

Extraction and ETL Differences

i) Structure: CSV contains flat, unformatted plain text. Excel uses structured, binary XML workbooks with multiple sheets.

ii) Parsing Speed: CSV extraction is extremely fast and low-memory. Excel requires heavy libraries to unpack files, slowing down ETL pipelines.

iii)Data Types: CSV treats all data as text strings. Excel stores data types explicitly like dates, numbers, or currencies.

iv) Schema Consistency: CSV forces a strict rows-and-columns layout. Excel allows scattered tables, empty rows, and merged cells that break automated extractors.

v) Data Volume: CSV has no row limit. Excel restricts files to 1,048,576 rows, making it unsuitable for large-scale data ingestion.

vi) Automation Friendliness: CSV works natively with any programming language. Excel often requires specific parsers to strip formulas and formatting.

Question 4 : Explain the steps involved in extracting data from a relational database.

Answer: To extract data from a relational database, an ETL pipeline connects to the source system, identifies data changes, and pulls the records while minimizing the performance impact on production environments.

The Extraction Process Steps

i) Establish Connection: The ETL tool connects to the database using specific drivers like JDBC or ODBC.

ii) Authenticate and Authorize: The pipeline logs in using dedicated credentials with read-only permissions for security.

iii) Define Target Scope: The system identifies specific tables, views, or schemas required for the data workload.

iv) Select Extraction Strategy: The pipeline determines whether to pull all data or only recent changes.

Full Extract: Queries and copies the entire table.

Incremental Extract: Uses timestamps or sequence keys to pull only new records.

Change Data Capture (CDC): Reads database transaction logs to capture changes in real-time.

v) Execute Queries: The tool runs optimized SQL statements, often using pagination to avoid overloading database memory.

vi) Stream and Stage Data: Extracted records are streamed into a temporary staging area or file format for the transformation phase.

Question 5 : Explain three common challenges faced during data extraction.

Answer: Data extraction faces significant hurdles due to unpredictable source systems, changing data structures, and the sheer volume of modern datasets.

Three Common Extraction Challenges

i) Schema Drift: Source systems frequently change their underlying data structures. When an application adds a new column, renames an existing field, or alters a data type, it immediately breaks downstream ETL extraction scripts and automated pipelines.

ii) Production Performance Impact: Querying massive amounts of data from live operational databases consumes heavy CPU and memory resources. If extraction processes are not carefully throttled or offloaded to replicas, they can severely slow down or crash applications used by active customers.

iii) Data Quality and Inconsistency: Source data is often messy, unstructured, or poorly validated. Extracting information across disparate systems frequently uncovers missing values, mismatched date formats, duplicate entries, and corrupted text encodings that complicate the ingestion phase.

Question 6 : What are APIs? Explain how APIs help in real-time data extraction.

Answer: An API (Application Programming Interface) is a software intermediary that allows two distinct applications to interact and share data with each other using a standardized set of rules and protocols.

How APIs Enable Real-Time Data Extraction

Unlike traditional batch extraction—which schedules large database dumps at midnight—APIs facilitate instantaneous, continuous data streaming through the following mechanisms:

i) Event-Driven Webhooks: Instead of the ETL pipeline constantly asking for updates, the source system uses a webhook to automatically push new data to the pipeline the exact millisecond an event occurs (e.g., a customer making a purchase).

ii) Continuous Polling Loops: Pipelines can be programmed to query lightweight API endpoints at high frequencies (such as every few seconds) to instantly grab the newest micro-batches of data.

iii) Low-Overhead Payloads: APIs typically exchange data in highly compressed formats like JSON or XML, allowing files to travel across networks instantly without causing bandwidth bottlenecks.

iv) System Decoupling: APIs act as a secure buffer. They let extraction tools grab real-time data from an application's interface without touching or slowing down the application's underlying live production database.

Question 7 : Why are databases preferred for enterprise-level data extraction?

Answer: Databases are preferred for enterprise-level data extraction because they provide the scalability, speed, and data integrity needed to handle massive corporate workloads safely.

Why Enterprises Prefer Databases

i) High Scalability: Databases handle petabytes of data effortlessly. Flat files break down under these massive enterprise volumes.

ii) Concurrent Access: Multiple ETL pipelines can extract data simultaneously. Built-in locking mechanisms prevent system crashes during these reads.

iii) Incremental Tracking: Databases support Change Data Capture (CDC) and timestamps. This allows pipelines to extract only new changes, saving time.

iv) Data Integrity: Strict schemas and ACID compliance ensure data remains uncorrupted. This guarantees high-quality data during the extraction phase.

v) Performance Optimization: Indexing and partitioning allow tools to locate records instantly. This makes queries faster and reduces extraction windows.

vi) Advanced Security: Access is managed through granular, role-based permissions. This ensures sensitive enterprise data is masked or restricted during extraction.

Question 8 : What steps should an ETL developer take when extracting data from large CSV files (1GB+)?

Answer: Extracting data from large CSV files (1GB+) requires specialized techniques to prevent the ETL pipeline from running out of system memory and crashing.

Essential Steps for Large CSV Extraction

i) Implement Chunking: Read the file in small, manageable pieces (e.g., 10,000 rows at a time) rather than loading the entire 1GB+ file into RAM at once.

ii) Optimize Data Types: Explicitly define column schemas and downcast data types (e.g., using 32-bit instead of 64-bit integers) during ingestion to minimize memory usage.

iii) Stream the Data: Use streaming APIs or generators to process the file line-by-line or block-by-block, keeping the memory footprint consistently low.

iv) Select Specific Columns: Drop or skip unnecessary columns at the initial extraction boundary to avoid processing redundant data.

v) Leverage Parallel Processing: Split the massive text file into smaller segments and utilize multi-core processing to extract multiple sections concurrently.

vi) Utilize Serverless Buffers: Stream the raw file directly into cloud staging areas (like Amazon S3 or Google Cloud Storage) before attempting complex parsing.
